As seen above, sometimes 10k events does not genorate any tau>150GeV at the lhe level. 

Strangely, these results seem to say three things:
1) the production of high pT taus are **primarially** from the *other* hadrons, not involved in the colation
2) 10k events is not suffishant to determin cross section if MadGraph is indeed doing this kind of lhe-level cut as sometimes there are no high-pT taus
3) MadGraph has repeatdly reported the same cross secion of .23pb across several 10k event runs; perhaps it has some other way of ensureing the cross section is stable, even when few taus pass the cut

Actually, I can test this third point, maybe the cut is running the simulation until it perduces 10k taus of sufishant pT. I can test this by looking at an lhe that had a MadGraph-level cut applied and check how many of the taus pass the cut. If all of them have pT>150GeV then MadGraph must be genorating taus until it gets 10k with sufishantly large pT. 

```python
taus_mg5cut = pd.read_csv("/home/leo/MadStats/MadGraph2/mg5amcnlo/MG5_CUT_NO_PITHIA/Events/run_02/taus_with_mg5_cut.dat", sep="\\s+", header=None)
```

```python
taus_mg5cut.columns = ['PID', 'Status', 'Parent1', 'Parent2', 'Color1', 'Color2', 'Px', 'Py', 'Pz', 'Energy', 'Mass', 'Lifetime', 'Spin']
```

```python
taus_mg5cut['pT'] = np.sqrt((taus_mg5cut['Px']**2)+(taus_mg5cut['Py']**2))
taus_mg5cut.shape[0]
```

```python
taus_mg5cut_150 = taus_mg5cut[taus_mg5cut['pT']>150]
taus_mg5cut_150.shape[0]
```

Aha, MadGraph must be continously genorating events until it finds 10k taus of pT>150GeV; this must be very compuationally expencive in some cases, especially when the post-cut cross section is very small. This is how it is able to get a numarically stable result, even when doing the same process in a naive way (at 10k events) generates as few as 0 taus that pass the cut.
