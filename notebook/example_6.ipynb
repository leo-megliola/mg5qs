### Weight results by luminosity

$$L=luminosity, N=number \space of \space events, \sigma=integrated \space cross \space section$$
$$L= \frac{N}{\sigma}$$
$$L_{max}=\frac{N}{\sigma_{min}}$$
$$L_{max}=\frac{N}{\sigma_{min}}=\frac{N}{w\sigma_i}$$
$$w=\frac{\sigma_{min}}{\sigma_{i}} \leq 1$$
**Therefore: keep all the events from the run with least cross section, and reduce events from all others by the above ratio.**

```python
sigma_min = sys.float_info.max # used to match weight across runs
gen_info = {}
for LHE in LHEs:
    params = qs.get_run_params(LHE)
    num_events, sigma_i, banner = qs.get_MGGenerationInfo(LHE)  # NOT INVERTING SIGMA 
    gen_info[banner] = sigma_i
    if sigma_i < sigma_min:
        sigma_min = sigma_i
```

```python
# reload at pTs from pickle files (should use qs.unpickle_pTs?)
path = Path('pT_example_6')
v = []
for fname in os.listdir(path):
    if 'pkl' in fname:
        with open(path / fname, 'rb') as f:
            v.append(pickle.load(f))
            
len(v), type(v[0][0]), type(v[0][1])
```

$$c(\alpha)=\sqrt{-\frac{1}{2}ln(\frac{\alpha}{2})}$$
and
$$D_{n,m}=c(\alpha)\sqrt{\frac{n+m}{nm}}$$

```python
c = lambda alpha: np.sqrt(-(1/2)*np.log(alpha/2))
levels = np.array([c(x) for x in [1-.6827, 1-.9545, 1-.9973, 1-.999937, 1-.9999994]])

D_nm = lambda c_alpha, n, m: c_alpha * np.sqrt((n+m)/(n*m))

print(levels)
print(D_nm(levels, 50000, 1000000))
```

```python
import scipy.stats as stats

params, _pTs = qs.unpickle_pTs('pT_example_4') # SM
BINS = int(np.ceil(np.max(_pTs)))
discrete_cdf = qs.DiscreteCDF(_pTs, BINS)

ks = []

for pT_val in v:  
    params, pTs = pT_val
    sigma_i = gen_info[params.file_spec]
    w = sigma_min / sigma_i
    N_prime = int(len(pTs[1]) * w)
    pTs_prime = pTs[1][0:N_prime-1]
    x = params.get_value('SMEFT', 971)
    y = params.get_value('SMEFTCPV', 820)
    z = stats.kstest(pTs_prime, discrete_cdf.cdf).statistic
    ks.append([x,y,z])

ks = np.array(ks)
```

```python
from scipy.interpolate import griddata
np.sort(ks, axis=1)
# Separate the x, y, and z values
x = ks[:, 0]
y = ks[:, 1]
z = ks[:, 2]

# Create grid points for x, y, and z
grid_x, grid_y = np.mgrid[-30:30:11j, -30:30:11j]

# Interpolate the z values on the grid (method can be 'linear', 'cubic', or 'nearest')
grid_z = griddata((x, y), z, (grid_x, grid_y), method='cubic')

# Compute contour levels 
c_alpha = np.array([c(x) for x in [1-.9545, 1-.9973, 1-.999937, 1-.9999994]])
levels = D_nm(c_alpha, 50000, 1000000)

# Create the contour plot
contour = plt.contour(grid_x, grid_y, grid_z, levels=levels, colors='black')

contour_labels = {levels[0]: '2 sigma', levels[1]: '3 sigma', levels[2]: '4 sigma', levels[3]: '5 sigma'}
fmt = {level: label for level, label in zip(levels, contour_labels.values())}
plt.clabel(contour, inline=True, fontsize=8, fmt=fmt)

formatted_IL = f"{sigma_min:.0f}"
plt.title(f"3332 v. SM, Integrated Luminocity = {formatted_IL} $pb^{{-1}}$")
plt.xlabel("cledqRe3332")
plt.ylabel("cledqIm3332")

# Show the plot
plt.show()
```
